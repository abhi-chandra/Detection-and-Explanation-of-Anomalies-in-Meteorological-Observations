```python
file_path_1 = '/content/drive/MyDrive/train_set_colab.csv'
file_path_2 = '/content/drive/MyDrive/test_set_colab.csv'
file_path_3 = '/content/drive/MyDrive/val_set_colab.csv'
```

```python
x1 = pd.read_csv(file_path_1, delimiter=',')
numeric_attributes = ['tmpc', 'dwpc','relh', 'wspeedkm','pressure','vsbykm', 'feelc']
df1= x1[numeric_attributes]
print(df1)
```

```python
x2 = pd.read_csv(file_path_2, delimiter=',')
numeric_attributes = ['tmpc', 'dwpc','relh', 'wspeedkm','pressure','vsbykm', 'feelc']
df2= x2[numeric_attributes]
print(df2)
```

```python
x3 = pd.read_csv(file_path_3, delimiter=',')
numeric_attributes = ['tmpc', 'dwpc','relh', 'wspeedkm','pressure','vsbykm', 'feelc']
df3= x3[numeric_attributes]
print(df3)
```

```python
dfs_to_concat = [df1, df2, df3]  # Add more DataFrames if needed

# Concatenate the DataFrames along the rows
df4 = pd.concat([df2, df3], ignore_index=True)
df5 = pd.concat(dfs_to_concat, ignore_index=True)
df6 = pd.concat([df1, df3], ignore_index=True)


# Print the new concatenated DataFrame df4
print(df6)
```

```python
from outliertree import OutlierTree

outliers_model = OutlierTree()
```

```python
outliers_df = outliers_model.fit(df5, outliers_print = 20, return_outliers = True)
```

```python
new_outliers = outliers_model.predict(df2)
```

```python
outliers_model.print_outliers(new_outliers)
```
